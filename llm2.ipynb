```python
from langchain_openai import ChatOpenAI

llm=ChatOpenAI(model="gpt-5.5")
```

```python
print(llm)
```

```python
result=llm.invoke("what is data engineering")
print(result)
```

```python
from langchain_core.prompts import ChatPromptTemplate
prompt=ChatPromptTemplate.from_messages([
    "system","You are an AI assistant, provide the required assistance to the questions asked",
    "user","{input}"
])
prompt
```

```python
chain=prompt|llm
response=chain.invoke({"input":"Can you explain to me what is linear algebra"})
print(response)
```

```python
#%pip install faiss-gpu
from langchain_community.vectorstores import FAISS

storedb=FAISS.from_documents(final_document,embeddings)

retriever=storedb.as_retriever()
```

```python
from langchain_classic.chains import create_retrieval_chain
from langchain_classic.chains.combine_documents import create_stuff_documents_chain

prompt2=ChatPromptTemplate.from_template(
    """
Answer the following question based only on the provided context:
<context>
{context}
</context>

Question:{input}
    """
)
```

```python
document_chain=create_stuff_documents_chain(llm,prompt2)
```

```python
document_chain
```

```python
retrieval_chain=create_retrieval_chain(retriever,document_chain)

response=retrieval_chain.invoke(
    {"input":"Clearly explain the palmer drought severity index, outlining the important facts",
     "context":"The Palmer Drought Severity Index (PDSI) is a widely used measure to assess and quantify drought conditions based on meteorological data"}
)

response['answer']
```
